```python
from pulp import *

# Define the Linear Programming problem
prob = LpProblem("Product Mix Problem", LpMaximize)

# Decision Variables
A = LpVariable("Product_A", lowBound=0, cat='Integer')
B = LpVariable("Product_B", lowBound=0, cat='Integer')

# Objective Function: Maximize Profit
prob += 25 * A + 20 * B, "Total Profit"

# Constraints
prob += 3 * A + 2 * B <= 100, "Machine Hours"
prob += 2 * A + 1 * B <= 80, "Labor Hours"

# Solve the problem
prob.solve()

# Print results
print("Status:", LpStatus[prob.status])
print("Optimal Production of Product A =", value(A))
print("Optimal Production of Product B =", value(B))
print("Maximum Profit = ₹", value(prob.objective))

# Optional: Save results to a file
with open("optimization_result.txt", "w") as f:
    f.write(f"Optimal Product A: {value(A)}\n")
    f.write(f"Optimal Product B: {value(B)}\n")
    f.write(f"Maximum Profit: ₹{value(prob.objective)}\n")
```
